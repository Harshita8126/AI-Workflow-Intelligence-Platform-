```python
# Recommendation Matcher
df_gaps = pd.read_csv(os.path.join(DATA_PROCESSED, "employee_skill_gaps.csv"))

recommendation_records = []
for _, emp in df_gaps.iterrows():
    emp_id = emp['employee_id']
    missing_list = emp['missing_skills_list'].split('|') if pd.notnull(emp['missing_skills_list']) and emp['missing_skills_list'] else []
    
    recommended = []
    for skill in missing_list[:3]:  # Top 3 priority missing skills
        # Match against catalog
        for _, c in df_courses.iterrows():
            if any(s.lower() in c['target_skills'].lower() or c['target_skills'].lower() in s.lower() for s in [skill]):
                recommended.append(f"{c['course_title']} ({c['course_id']})")
                break
        if len(recommended) >= 2:
            break
            
    if not recommended:
        recommended.append("Critical Thinking & Complex Decision Making (C-SOFT-01)")
        
    recommendation_records.append({
        'employee_id': emp_id,
        'recommended_learning_plan': " + ".join(recommended)
    })

df_recs = pd.DataFrame(recommendation_records)
out_recs = os.path.join(DATA_PROCESSED, "employee_recommendations.csv")
df_recs.to_csv(out_recs, index=False)
print("=== SAMPLE RECOMMENDATIONS ===")
print(df_recs.head(5).to_string(index=False))
```
