### Purpose:
Compute actual organization-wide and department-level engagement, attendance, KPI, and performance scores.


```python
import os
import pandas as pd
import numpy as np

DATA_PROCESSED = "../data/processed"
df_perf = pd.read_csv(os.path.join(DATA_PROCESSED, "engagement_processed.csv"))

print("=== OVERALL WORKFORCE METRICS ===")
print(f"Total Employees: {len(df_perf):,}")
print(f"Average Engagement Score: {df_perf['engagement_score'].mean():.2f}%")
print(f"Average Performance Score: {df_perf['performance_score'].mean():.2f}")
print(f"Average KPI Score: {df_perf['kpi_score'].mean():.2f}%")
print(f"Average Attendance: {df_perf['attendance_pct'].mean():.2f}%")
```

```python
# Department Breakdown
dept_summary = df_perf.groupby('department').agg(
    employee_count=('employee_id', 'count'),
    avg_engagement=('engagement_score', 'mean'),
    avg_performance=('performance_score', 'mean'),
    avg_kpi=('kpi_score', 'mean'),
    avg_attendance=('attendance_pct', 'mean'),
    promotion_eligible=('promotion_eligibility', lambda x: (x == 'Yes').sum())
).round(2).reset_index()

print("=== DEPARTMENT ENGAGEMENT SUMMARY ===")
print(dept_summary.to_string(index=False))

dept_summary_file = os.path.join(DATA_PROCESSED, "department_engagement_summary.csv")
dept_summary.to_csv(dept_summary_file, index=False)
print(f"Saved: {dept_summary_file}")
```
