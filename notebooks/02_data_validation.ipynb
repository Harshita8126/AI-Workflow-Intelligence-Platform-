# 02 Data Validation
**Enterprise HR AI — Workforce Intelligence & Upskilling Platform**

### Objective:
Perform rigorous schema, type, range, uniqueness, and categorical validation across all raw datasets.


```python
import os
import pandas as pd
import numpy as np

DATA_RAW = "../data/raw"
df_attrition = pd.read_csv(os.path.join(DATA_RAW, "employee_attrition.csv"))
df_perf = pd.read_csv(os.path.join(DATA_RAW, "hr_performance_engagement.csv"))
df_occ = pd.read_csv(os.path.join(DATA_RAW, "occupation_data.csv"))
df_ess = pd.read_csv(os.path.join(DATA_RAW, "essential_skills.csv"))
df_soft = pd.read_csv(os.path.join(DATA_RAW, "software_skills.csv"))

print("Datasets loaded for validation.")
```

```python
# 1. Validation: Attrition Dataset
print("Validating employee_attrition.csv...")
assert df_attrition['Age'].between(18, 100).all(), "Validation Error: Age out of bounds"
assert df_attrition['EmployeeNumber'].is_unique, "Validation Error: Duplicate EmployeeNumber"
assert set(df_attrition['Attrition'].unique()) <= {'Yes', 'No'}, "Validation Error: Invalid Attrition values"
assert df_attrition['MonthlyIncome'].min() > 0, "Validation Error: MonthlyIncome must be positive"
assert df_attrition['TotalWorkingYears'].min() >= 0, "Validation Error: TotalWorkingYears cannot be negative"
assert df_attrition['YearsAtCompany'].min() >= 0, "Validation Error: YearsAtCompany cannot be negative"
print("Passed employee_attrition.csv validation checks.")
```

```python
# 2. Validation: Performance & Engagement Dataset
print("Validating hr_performance_engagement.csv...")
assert df_perf['Employee ID'].is_unique, "Validation Error: Duplicate Employee ID in performance data"
assert df_perf['Performance Score'].between(0, 100).all(), "Validation Error: Performance Score out of range"
assert df_perf['KPI Score'].between(0, 100).all(), "Validation Error: KPI Score out of range"
assert df_perf['Attendance (%)'].between(0, 100).all(), "Validation Error: Attendance out of range"
assert df_perf['Task Completion (%)'].between(0, 100).all(), "Validation Error: Task Completion out of range"
assert df_perf['Peer Rating'].between(1.0, 5.0).all(), "Validation Error: Peer Rating out of range [1, 5]"
assert df_perf['Manager Feedback'].between(1.0, 5.0).all(), "Validation Error: Manager Feedback out of range [1, 5]"
assert set(df_perf['Promotion Eligibility'].unique()) <= {'Yes', 'No'}, "Validation Error: Invalid Promotion Eligibility"
print("Passed hr_performance_engagement.csv validation checks.")
```

```python
# 3. Validation: O*NET Occupation and Skills Taxonomies
print("Validating occupation and skills taxonomies...")
assert df_occ['O*NET-SOC Code'].is_unique, "Validation Error: Duplicate SOC code in occupation_data"
assert df_occ['Title'].notnull().all(), "Validation Error: Missing occupation title"
assert set(df_ess['O*NET-SOC Code']).issubset(set(df_occ['O*NET-SOC Code'])), "Validation Error: Unmapped SOC codes in essential_skills"
assert set(df_soft['O*NET-SOC Code']).issubset(set(df_occ['O*NET-SOC Code'])), "Validation Error: Unmapped SOC codes in software_skills"
assert df_ess['Data Value'].between(0.0, 7.0).all(), "Validation Error: Essential skill scale value out of range"
print("Passed occupation_data, essential_skills, and software_skills integrity checks.")
```
