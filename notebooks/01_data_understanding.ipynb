### Objective:
Load and inspect all raw datasets in `data/raw/` to examine shapes, columns, data types, missing values, duplicates, target distribution, and entity characteristics.


```python
import os
import pandas as pd
import numpy as np

DATA_RAW = "../data/raw"
files = sorted([f for f in os.listdir(DATA_RAW) if f.endswith('.csv')])
print("Raw CSV files found in data/raw/:")
for f in files:
    size = os.path.getsize(os.path.join(DATA_RAW, f))
    print(f" - {f} ({size:,} bytes)")
```

```python
def profile_dataset(name, filename):
    path = os.path.join(DATA_RAW, filename)
    df = pd.read_csv(path)
    print("=" * 75)
    print(f"DATASET: {name} ({filename})")
    print("=" * 75)
    print(f"Shape: {df.shape[0]} rows x {df.shape[1]} columns")
    print(f"Duplicates: {df.duplicated().sum()}")
    print("\nColumns, Dtypes, and Missing Values:")
    col_summary = pd.DataFrame({
        'Dtype': df.dtypes,
        'Null_Count': df.isnull().sum(),
        'Null_%': (df.isnull().sum() / len(df) * 100).round(2),
        'Uniques': df.nunique()
    })
    print(col_summary.to_string())
    print("\nSample Top 2 Rows:")
    print(df.head(2).to_dict(orient='records'))
    return df

df_attrition = profile_dataset("Employee Attrition", "employee_attrition.csv")
df_perf = profile_dataset("HR Performance & Engagement", "hr_performance_engagement.csv")
df_occ = profile_dataset("Occupation Master", "occupation_data.csv")
df_ess = profile_dataset("Essential Skills", "essential_skills.csv")
df_soft = profile_dataset("Software Skills", "software_skills.csv")
```

```python
print("Attrition Target Breakdown:")
print(df_attrition['Attrition'].value_counts(normalize=True).mul(100).round(2).to_string())
```
