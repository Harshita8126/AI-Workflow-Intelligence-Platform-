# 05 Feature Engineering
**Enterprise HR AI — Workforce Intelligence & Upskilling Platform**

### Purpose:
Engineer domain-grounded predictive features, apply ethical governance (exclude sensitive demographics), and prepare model-ready train/test feature sets.


```python
import os
import pandas as pd
import numpy as np

DATA_PROCESSED = "../data/processed"
df = pd.read_csv(os.path.join(DATA_PROCESSED, "employee_attrition_processed.csv"))
print(f"Loaded processed attrition data: {df.shape[0]} rows x {df.shape[1]} cols")
```

```python
# 1. Feature Engineering with Business & Behavioral Rationale
# A. Loyalty Ratio: Proportion of total working career spent at this company
df['tenure_loyalty_ratio'] = np.round(df['years_at_company'] / (df['total_working_years'] + 1), 4)

# B. Career Stagnation Index: Years without promotion relative to tenure in current role
df['stagnation_index'] = np.round(df['years_since_last_promotion'] / (df['years_in_current_role'] + 1), 4)

# C. Composite Morale & Satisfaction: Average of 4 core satisfaction Likert scores
df['composite_satisfaction'] = np.round(
    (df['job_satisfaction'] + df['environment_satisfaction'] + 
     df['relationship_satisfaction'] + df['job_involvement']) / 4.0, 
    2
)

# D. Income Acceleration: Monthly income relative to total working years
df['income_per_working_year'] = np.round(df['monthly_income'] / (df['total_working_years'] + 1), 2)

print("Engineered 4 new domain features:")
print(df[['tenure_loyalty_ratio', 'stagnation_index', 'composite_satisfaction', 'income_per_working_year']].head())
```

```python
# 2. Predictive Governance: Exclude sensitive demographic & constant attributes
SENSITIVE_ATTRS = ['gender', 'marital_status']
TARGET_COLS = ['attrition', 'attrition_binary']
ID_COLS = ['employee_id']

feature_cols = [c for c in df.columns if c not in SENSITIVE_ATTRS + TARGET_COLS + ID_COLS]
print(f"Total predictive features selected ({len(feature_cols)}):")
print(feature_cols)

X = df[feature_cols]
y = df['attrition_binary']

# Save engineered features
engineered_file = os.path.join(DATA_PROCESSED, "attrition_features_engineered.csv")
df.to_csv(engineered_file, index=False)
print(f"Saved engineered dataset to: {engineered_file}")
```
